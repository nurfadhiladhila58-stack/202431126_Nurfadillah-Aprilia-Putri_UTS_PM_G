```python
# nurfadillahAP_202431126

x = data.drop(['round_winner'], axis=1)
y = data['round_winner']

x.head()
```

```python
#nurfadillahAP_202431126
y.head()
```

```python
#nurfadillahAP_202431126
x = pd.get_dummies(x, columns=['map'], dtype=int)

x['bomb_planted'] = x['bomb_planted'].astype(int)

print(x.dtypes)
x.head()
```

```python
#nurfadillahAP_202431126

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25, random_state=0
)
```

```python
# nurfadillahAP_202431126
print(f'x_train shape : {x_train.shape}')
print(f'x_test shape : {x_test.shape}')
```

```python
#nurfadillahAP_202431126

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaler.fit(x_train)

x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# nurfadillahAP_202431126

from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=7)

knn.fit(x_train_scaled, y_train)
```

```python
#nurfadillahAP_202431126
y_pred = knn.predict(x_test_scaled)

y_pred
```

```python
#nurfadillahAP_202431126

from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score

accuracy_k7 = accuracy_score(y_test, y_pred)

print('Akurasi KNN (K=7):', accuracy_k7)
print(classification_report(y_test, y_pred))
```

```python
#nurfadillahAP_202431126

K_TARGET = 15

model_k15 = KNeighborsClassifier(n_neighbors=K_TARGET)

model_k15.fit(x_train_scaled, y_train)

y_pred_k15 = model_k15.predict(x_test_scaled)

accuracy_k15 = accuracy_score(y_test, y_pred_k15)

print(f"Nilai K yang digunakan: {K_TARGET}")
print(f"Akurasi Model KNN pada data test: {accuracy_k15:.4f}")
```

```python
#nurfadillahAP_202431126
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import make_pipeline

k_values = range(1, 20)
accuracies = []

for k in k_values:
    model = make_pipeline(
        StandardScaler(),
        KNeighborsClassifier(n_neighbors=k)
    )

    scores = cross_val_score(
        model, x_train, y_train, cv=5, scoring='accuracy'
    )

    accuracies.append(scores.mean())

plt.figure(figsize=(8, 5))

plt.plot(k_values, accuracies, marker='o')

plt.xlabel('Nilai K')
plt.ylabel('Rata-rata Akurasi Cross Validation')
plt.title('Pemilihan Nilai K pada KNN')
plt.grid(True)
plt.show()

print("Nilai K terbaik:", list(k_values)[np.argmax(accuracies)])
```

```python
#nurfadillahAP_202431126

from sklearn.naive_bayes import GaussianNB

model_nb = GaussianNB()

model_nb.fit(x_train_scaled, y_train)

y_pred_nb = model_nb.predict(x_test_scaled)

accuracy_nb = accuracy_score(y_test, y_pred_nb)

print('Akurasi Naive Bayes:', accuracy_nb)
print(classification_report(y_test, y_pred_nb))
```

```python
#nurfadillahAP_202431126
hasil = pd.DataFrame({
    'Algoritma': ['KNN (K=7)', 'KNN (K=15)', 'Naive Bayes'],
    'Akurasi': [accuracy_k7, accuracy_k15, accuracy_nb]
})

hasil['Akurasi (%)'] = hasil['Akurasi'] * 100

hasil
```

```python
#nurfadillahAP_202431126

plt.figure(figsize=(8, 5))

plt.bar(hasil['Algoritma'], hasil['Akurasi (%)'])

plt.ylabel('Akurasi (%)')
plt.xlabel('Algoritma')
plt.title('Perbandingan Akurasi KNN dan Naive Bayes')
plt.ylim(0, 100)

plt.show()
```
